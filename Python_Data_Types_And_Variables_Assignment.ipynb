## Question 10: Dynamic Typing

Use the 'type()' function to observe how the data type of 'x' changes after different operations.

```python
x = 4
print(x)
type(x)
```

### 4 is a whole number, so Python automatically identifies it as an integer (int).

```python
x += 1
print(x)
type(x)
```

### x += 1 changes the value from 4 to 5, but because 5 is still a whole number, the data type remains int.

```python
x += 1.0
print(x)
type(x)
```

### After adding 1.0, the value becomes 6.0 and the data type changes to float.

This demonstrates Python’s dynamic typing, where a variable can hold values of different data types during a program.
